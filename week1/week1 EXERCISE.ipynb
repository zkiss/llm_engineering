````python
# imports
import os
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from openai import OpenAI
````

````python
# constants
MODEL_GPT = 'gpt-4o-mini'
MODEL_OLLAMA = 'gemma3:1b'
````

````python
# set up environment
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')

if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")
elif not api_key.startswith("sk-proj-"):
    print("An API key was found, but it doesn't start sk-proj-; please check you're using the right key - see troubleshooting notebook")
elif api_key.strip() != api_key:
    print("An API key was found, but it looks like it might have space or tab characters at the start or end - please remove them - see troubleshooting notebook")
else:
    print("API key found and looks good so far!")
````

````python
# Use OpenAI
openai = OpenAI()
MODEL=MODEL_GPT
````

````python
# Use local ollama
openai = OpenAI(base_url='http://localhost:11434/v1', api_key='ollama')
MODEL=MODEL_OLLAMA
````

````python
def answer(question):
    system_prompt="""\
    You are a helpful assistant with deep technical knowledge, helping engineers understanding various topics in depth.
    Yout task is to answer the questions presented to you by the user.
    Always start by providing a brief high level overview / executive summary.
    After that, expand on the topic in the style of a wikipedia article.
    Your answer should be technically accurate, informative, and should highlight any further areas to look into that are relevant to the original query (eg see also).
    Produce the answer in markdown.
    """
    stream = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": question}
        ],
        stream=True
    )
    response = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        response += chunk.choices[0].delta.content or ''
        response = response.replace("```","").replace("markdown", "")
        update_display(Markdown(response), display_id=display_handle.display_id)
````

````python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
yield from {book.get("author") for book in books if book.get("author")}
"""
````

````python
answer("""
Participants with what experience would be very useful to add to a conversation I'm setting up with the following instructions:

---
You are collaborating with others in a chat room.
The problem you're trying to solve is figuring out what changes are needed in society
and in the political system to make life more fair;
eliminating regulatory capture, and only allow elected individuals and teams they appoint to
influence the rules of the system.

When prompted, you have two options:
1: LISTEN: say nothing and keep listening to the conversation
2: MESSAGE: send a message to the chat room that everybody else will receive

Your output should either be:
LISTEN

or

MESSAGE
<your message, can be multiline>
---

List the participant's role, and a system prompt that describes their role and expertise,
assuming the above system prompt will be appended after the role specific portion of the prompt
you suggest (so you don't have to).
""")
````

````python
# Get answer, with streaming
answer(question)
````
